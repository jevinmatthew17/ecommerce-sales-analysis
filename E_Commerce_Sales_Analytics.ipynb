```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Create synthetic E-commerce data
np.random.seed(42)
data_size = 5000
data = {
    'order_id': np.arange(1001, 1001 + data_size),
    'date': [datetime(2025, 1, 1) + timedelta(days=np.random.randint(0, 365)) for _ in range(data_size)],
    'product_category': np.random.choice(['Electronics', 'Clothing', 'Home & Kitchen', 'Beauty', 'Sports'], data_size),
    'sales_amount': np.random.uniform(10, 500, data_size).round(2),
    'quantity': np.random.randint(1, 6, data_size),
    'payment_method': np.random.choice(['Credit Card', 'PayPal', 'UPI', 'Debit Card'], data_size),
    'customer_rating': np.random.randint(1, 6, data_size)
}

df_eco = pd.DataFrame(data)

# Inject "dirty" data for cleaning practice
df_eco.iloc[0:10, 2] = None  # Missing categories
df_eco = pd.concat([df_eco, df_eco.iloc[0:5]], ignore_index=True) # Duplicates

print("✅ E-commerce Data Generated!")
```

```python
# 1. Remove Duplicates
df_eco = df_eco.drop_duplicates()

# 2. Handle Missing Values
df_eco['product_category'] = df_eco['product_category'].fillna('Unknown')

# 3. Feature Engineering (Crucial for Dashboards)
df_eco['month'] = df_eco['date'].dt.strftime('%b')
df_eco['day_of_week'] = df_eco['date'].dt.day_name()
df_eco['total_revenue'] = df_eco['sales_amount'] * df_eco['quantity']

# 4. Export
df_eco.to_csv('ecommerce_cleaned.csv', index=False)
from google.colab import files
files.download('ecommerce_cleaned.csv')
print("✅ Data Cleaned & Downloaded!")
```
